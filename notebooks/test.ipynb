```python
from src.model import UNet
import torch
tensor = torch.randn(1, 4, 240, 240)
model = UNet(in_channels=4, out_channels=4)  # UNet model
output = model(tensor)
print(f"Model output shape: {output.shape}")
```

```python
import torch
import torch.nn as nn
from src.train import train

train_test_split = 0.8
dataset_size = len(dataset)
train_size = int(train_test_split * dataset_size)
test_size = dataset_size - train_size
train_dataset, test_dataset = torch.utils.data.random_split(dataset, [train_size, test_size])

print(f"Train dataset size: {len(train_dataset)}")
print(f"Test dataset size: {len(test_dataset)}")

train(train_dataset, test_dataset, model, num_epochs=10, batch_size=4, learning_rate=1e-3)
```
