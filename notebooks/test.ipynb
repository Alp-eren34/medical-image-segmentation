```python
root = r"D:\Projects\medical-image-segmentation\data\MICCAI_BraTS2020_TrainingData"
tum_hastalar = []

for hasta_klasoru in os.listdir(root):
    hasta_yolu = os.path.join(root, hasta_klasoru)
    if not os.path.exists(os.path.join(hasta_yolu, f"{hasta_klasoru}_flair.nii")) or \
       not os.path.exists(os.path.join(hasta_yolu, f"{hasta_klasoru}_seg.nii")) or \
       not os.path.exists(os.path.join(hasta_yolu, f"{hasta_klasoru}_t1.nii")) or \
       not os.path.exists(os.path.join(hasta_yolu, f"{hasta_klasoru}_t1ce.nii")) or \
       not os.path.exists(os.path.join(hasta_yolu, f"{hasta_klasoru}_t2.nii")):
        continue
    if os.path.isdir(hasta_yolu):
        hasta_verisi = {
            'hasta_id': hasta_klasoru,
            'flair': os.path.join(hasta_yolu, f"{hasta_klasoru}_flair.nii"),
            'seg': os.path.join(hasta_yolu, f"{hasta_klasoru}_seg.nii"),
            't1': os.path.join(hasta_yolu, f"{hasta_klasoru}_t1.nii"),
            't1ce': os.path.join(hasta_yolu, f"{hasta_klasoru}_t1ce.nii"),
            't2': os.path.join(hasta_yolu, f"{hasta_klasoru}_t2.nii")
        }
        tum_hastalar.append(hasta_verisi)
print(f"Toplam {len(tum_hastalar)} hastanın klasörü bulundu ve listeye eklendi.\n")

'''
for hasta in tum_hastalar:
    try:
        print(f'flair shape: {nib.load(hasta["flair"]).get_fdata().shape}')
        print(f'seg shape: {nib.load(hasta["seg"]).get_fdata().shape}')
        print(f't1 shape: {nib.load(hasta["t1"]).get_fdata().shape}')
        print(f't1ce shape: {nib.load(hasta["t1ce"]).get_fdata().shape}')
        print(f't2 shape: {nib.load(hasta["t2"]).get_fdata().shape}')
    except Exception as e:
        print(f"Hata oluştu: {e}")
'''
```

```python
# Initialize pixel count array for 5 classes (0, 1, 2, 3, 4)
# Class 4 will be merged with class 3 later (as done in dataset.py)
pixel_counts = [0, 0, 0, 0, 0]

# Iterate over all patients in the dataset
for hasta in tum_hastalar:
    
    # Load the segmentation mask for the current patient
    seg = nib.load(hasta["seg"]).get_fdata()
    
    # Get unique label values and their pixel counts in this segmentation mask
    values, counts = np.unique(seg, return_counts=True)
    
    # Accumulate pixel counts for each class across all patients
    for v, c in zip(values, counts):
        pixel_counts[int(v)] += c
```
